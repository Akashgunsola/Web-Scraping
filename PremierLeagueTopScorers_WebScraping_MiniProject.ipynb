### Objective of this Project
- scrape data from a public website
- organize the data as a dataframe
- export the data as an excel sheet


```python
#Importong Libraries

import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
#Website which is to be scraped for
url = "https://www.bbc.com/sport/football/premier-league/top-scorers"
```

```python
try:
    response = requests.get(url)
    response.raise_for_status()
except Exception as e:
    print(e)
else:
    soup = BeautifulSoup(response.content,"html.parser")
    print("Succesful!")
```

```python
players = soup.find("tbody").find_all("tr", class_ = "ssrcss-qqhdqi-TableRowBody e1icz100")
playerName = []
goalsScored = []
teamName = []
assists = []
matchesPlayed = []
total_shots = []
```

```python
for player in players:
    names = player.find("div", class_ = "ssrcss-m6ah29-PlayerName e1n8xy5b1").get_text(strip = True)
    goals = int(player.find("div", class_ = "ssrcss-18ap757-CellWrapper ef9ipf0").get_text(strip=True))
    team_name = player.find("div", class_ = "ssrcss-qvpga1-TeamsSummary e1n8xy5b0").get_text(strip = True)
    stats = player.find_all("div", class_ = "ssrcss-1vo7v3r-CellWrapper ef9ipf0")
    assist = int(stats[0].get_text(strip=True))
    matches_Played = int(stats[3].get_text(strip=True))
    total_shot = int(stats[-3].get_text(strip=True))
    
    playerName.append(names)
    goalsScored.append(goals)
    teamName.append(team_name)
    assists.append(assist)
    matchesPlayed.append(matches_Played)
    total_shots.append(total_shot)

data = {
    'player' : playerName,
    'team' : teamName,
    'matches': matchesPlayed,
    'goals' : goalsScored,
    'assists' : assists,
    'shots' : total_shots
}

df_players = pd.DataFrame(data)
```
